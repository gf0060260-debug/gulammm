```python
x = file.drop(columns=["Loan_Status","Loan_ID"],axis=1)
```

```python
y = file["Loan_Status"]
```

```python
y = y.map({"N":0,"Y":1})
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer,make_column_selector
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,classification_report,recall_score,confusion_matrix
```

```python
x_train,x_test , y_train,y_test=train_test_split(x,y, test_size=0.2,random_state=42)
```

```python
num_colums = Pipeline([
    ("num",SimpleImputer(strategy="median")),
    ("scaling",StandardScaler())
])
cat_colums = Pipeline([
    ("cat",SimpleImputer(strategy="most_frequent")),
    ("encoding",OneHotEncoder(handle_unknown="ignore",sparse=False))
])
```

```python
transformer = ColumnTransformer([
    ("num_column",num_colums,make_column_selector(dtype_include=np.number)),
     ("cat_column",cat_colums,make_column_selector(dtype_include="object"))
])
```

```python
model_pipeline = Pipeline([
    ("preprocessing",transformer),
    ("model",LogisticRegression())
])
```

```python
model_pipeline
```

```python
model_pipeline.fit(x_train,y_train)
```

# training data predict

```python
y_predict_train = model_pipeline.predict(x_train)
```

```python
y_predict_train
```

# actual and predict value1

```python
pd.DataFrame({
    "actual":y_train,
    "predicted value":y_predict_train
})
```

# testing data predicted

```python
y_predict_test =model_pipeline.predict(x_test)
```

```python
y_predict_test
```

# actual and predicted value

```python
pd.DataFrame({
    "actual":y_test,
    "predicted":y_predict_test
})
```

```python
accuracy_score(y_test,y_predict_test)
```

```python
confusion_matrix(y_test,y_predict_test)
```
